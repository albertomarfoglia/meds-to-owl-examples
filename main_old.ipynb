```python
import yaml
import os
import polars as pl

MIMIC_COHORT = "MIMIC/MEDS_cohort"
TIME_OPT = "TS"


def init_dirs(task_name: str, index: int):
    export_dir = f"exports/{task_name}"
    graph_dir = f"{export_dir}/graph"
    outcomes_dir = f"{graph_dir}/labels"
    meds_cohort_dir = f"{export_dir}/meds/{index}/MEDS_cohort"
    os.makedirs(export_dir, exist_ok=True)
    os.makedirs(graph_dir, exist_ok=True)
    os.makedirs(outcomes_dir, exist_ok=True)
    os.makedirs(meds_cohort_dir, exist_ok=True)
    os.makedirs(f"{export_dir}/meds/{index}/MEDS_cohort/metadata", exist_ok=True)

    return graph_dir, outcomes_dir, meds_cohort_dir


with open("MIMIC/experiments.yaml", "r") as f:
    config = yaml.safe_load(f)

events = pl.scan_parquet(f"{MIMIC_COHORT}/data/**/*.parquet", low_memory=True).select(
    "subject_id", "time", "code", "numeric_value", "text_value"
)
```

```python
from MIMIC.utils.preprocessing import *
import numpy as np
import joblib
from meds2rdf import MedsRDFConverter, NTriplesSink, Config, MEDSSchema

columns = [
    "subject_id",
    "code",
    "time",
    "numeric_value",
    "prediction_time",
    "boolean_value",
]

for group_name, experiments in config["experiments"].items():
    print(f"\n=== Group: {group_name} ===")

    for scfg in experiments:
        print(f"Running task: {scfg['task']}")

        # if scfg["task"] != "first_24_in_hospital_mortality":
        #     continue

        IHM = f"{MIMIC_COHORT}/labels/{scfg['task']}/**/*.parquet"
        outcomes = pl.scan_parquet(IHM).select(
            "subject_id", "prediction_time", "boolean_value"
        )

        samples = generate_samples(
            labels=outcomes,
            n_folds=scfg["num_of_samples"],
            size=scfg["sample_size"],
            seed=1234,
            low_true_values=scfg["fixed_true"],
        )

        for index, sample in enumerate(samples):
            graph_dir, outcomes_dir, meds_cohort_dir = init_dirs(scfg["task"], index)

            events_base = (
                events.join(sample.lazy(), on="subject_id", how="inner")
                .with_columns(remove_long_text)
                .with_columns(birthdate_to_age)
                .filter(window_dict[scfg["window"]])
            )

            events_base = (
                # events_base.join(freq_codes_v(events_base, freq=100), on="code", how="semi")
                events_base.with_columns(parsed=parse_txt_to_float)
                .with_columns(swap_text_with_numeric)
                .with_columns(orig_code=pl.col("code"), orig_text=pl.col("text_value"))
                .with_columns(bp_values=bp_values_list)
                .explode("bp_values")
                .with_columns(numeric_value=pl.col("bp_values"))
                .with_columns(code=to_systolic_and_dyastolic_pressure)
                .with_columns(text_value=set_BP_values_text_to_null)
                .drop(derived_columns)
                .with_columns(regenerate_ids)
            ).select(columns)

            # events_base = aggregate_events(events_base, agg=scfg["agg"]).collect(engine="streaming")

            TRESHOLD = (
                scfg["sample_size"] / 100
            )  # max(1, int(np.ceil(scfg["sample_size"] * 0.00)))

            static_events = (
                events_base.filter(
                    pl.col("time").is_null() & pl.col("numeric_value").is_null()
                )
                .sort(["subject_id", "code"])
                .unique(subset=["subject_id", "code"], keep="last")
            )

            static_numeric_events = (
                events_base.filter(
                    pl.col("time").is_null() & pl.col("numeric_value").is_not_null()
                )
                .sort(["subject_id", "code"])
                .unique(subset=["subject_id", "code"], keep="last")
            )

            dynamic_numeric = (
                events_base.filter(
                    pl.col("time").is_not_null() & pl.col("numeric_value").is_not_null()
                )
                .with_columns(pl.col("time").dt.truncate(scfg["agg"]).alias("time"))
                .sort(["subject_id", "code", "time"])
                .group_by(["subject_id", "code", "time"])
                .agg(
                    [
                        # pl.col("numeric_value").mean().alias("mean"),
                        pl.col("numeric_value").min().alias("min"),
                        pl.col("numeric_value").max().alias("max"),
                        # pl.col("text_value").first(),
                        pl.col("prediction_time").last(),
                        pl.col("boolean_value").last(),
                    ]
                )
            )

            EPS = 1e-6

            # dynamic_numeric = dynamic_numeric.with_columns(
            #     [
            #         pl.when((pl.col("min") - pl.col("mean")).abs() > EPS)
            #         .then(pl.col("min"))
            #         .otherwise(None)
            #         .alias("min"),
            #         pl.when((pl.col("max") - pl.col("mean")).abs() > EPS)
            #         .then(pl.col("max"))
            #         .otherwise(None)
            #         .alias("max"),
            #     ]
            # )

            dynamic_numeric = (
                dynamic_numeric.unpivot(
                    index=[
                        "subject_id",
                        "code",
                        "time",
                        # "text_value",
                        "prediction_time",
                        "boolean_value",
                    ],
                    # on=["min", "mean", "max"],
                    on=["min", "max"],
                    variable_name="stat",
                    value_name="numeric_value",
                )
                .with_columns(
                    pl.concat_str([pl.col("code"), pl.lit("_"), pl.col("stat")]).alias(
                        "code"
                    )
                )
                .drop("stat")
            ).filter(pl.col("numeric_value").is_not_null())

            dynamic_numeric_codes = (
                dynamic_numeric.group_by(["subject_id", "code"])
                .len()
                .group_by("code")
                .len()
                .filter(pl.col("len") > TRESHOLD)
            )

            dynamic_codes = (
                events_base.filter(
                    pl.col("time").is_not_null() & pl.col("numeric_value").is_null()
                )
                # .group_by(["subject_id", "code"])
                # .agg(
                #     [
                #         pl.col("time").last().alias("time"),
                #         pl.len().cast(pl.Float64).alias("numeric_value"),
                #         pl.col("prediction_time").last().alias("prediction_time"),
                #         pl.col("boolean_value").last().alias("boolean_value"),
                #     ]
                # )
            )  # aggregare i codici per frequenza per paziente ed escludere quelli sotto una certa soglia (10%)

            dynamic_code_codes = (
                dynamic_codes.group_by(["subject_id", "code"])
                .len()
                .group_by("code")
                .len()
                .filter(pl.col("len") > TRESHOLD)
            )

            # dynamic_codes = dynamic_codes.join(
            #     dynamic_code_codes.select("code"), on="code", how="inner"
            # )

            final_events = pl.concat(
                [
                    static_numeric_events.select(columns),
                    dynamic_numeric.select(columns),
                    dynamic_codes.select(columns),
                    static_events.select(columns),
                ],
            )

            final_events = aggregate_events(final_events, agg=scfg["agg"]).collect(
                engine="streaming"
            )

            print(f"TOTAL EVENTS: {len(final_events)}")
            print(f"TOTAL CODES: {len(final_events.group_by('code').len())}")

            (_, _, split_l) = create_meds_cohort(
                final_events,
                orig_dir=MIMIC_COHORT,
                output_dir=meds_cohort_dir,
                columns=["subject_id", "code", "time", "numeric_value"],
            )

            joblib.dump(
                value=split_l.sort("subject_id")["boolean_value"].to_numpy(),
                filename=f"{outcomes_dir}/outcomes_meds_{TIME_OPT}_{scfg['sample_size']}_{index}.joblib",
            )

            MedsRDFConverter(meds_cohort_dir).convert(
                sink=NTriplesSink(
                    Path(f"{graph_dir}/meds_{scfg['sample_size']}_{index}"),
                    gzip_mode=False,
                ),
                cfg=Config(schemas={MEDSSchema.CODES}),
            )
```

```python
import pandas as pd
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score, accuracy_score
from utils import build_features

ROOT = "exports/first_24_in_hospital_mortality/meds/0/MEDS_cohort"

df1 = build_features(pl.read_parquet(f"{ROOT}/data/train/0.parquet"))
df2 = build_features(pl.read_parquet(f"{ROOT}/data/tuning/0.parquet"))
df3 = build_features(pl.read_parquet(f"{ROOT}/data/held_out/0.parquet"))

l1 = pl.read_parquet(f"{ROOT}/labels/train/0.parquet")
l2 = pl.read_parquet(f"{ROOT}/labels/tuning/0.parquet")
l3 = pl.read_parquet(f"{ROOT}/labels/held_out/0.parquet")

x_train = (
    df1.sort(pl.col("subject_id")).to_pandas().select_dtypes(exclude=["datetime64[ns]"])
)
y_train = l1.sort(pl.col("subject_id"))["boolean_value"]

X_tuning = (
    df2.sort(pl.col("subject_id")).to_pandas().select_dtypes(exclude=["datetime64[ns]"])
)
y_tuning = l2.sort(pl.col("subject_id"))["boolean_value"]

X_held = (
    df3.sort(pl.col("subject_id")).to_pandas().select_dtypes(exclude=["datetime64[ns]"])
)
y_held = l3.sort(pl.col("subject_id"))["boolean_value"]

x_train = pd.get_dummies(x_train).drop(columns=["subject_id"])
X_tuning = (
    pd.get_dummies(X_tuning)
    .drop(columns=["subject_id"])
    .reindex(columns=x_train.columns, fill_value=0)
)
X_held = (
    pd.get_dummies(X_held)
    .drop(columns=["subject_id"])
    .reindex(columns=x_train.columns, fill_value=0)
)

feature_names = x_train.columns.to_list()

model = XGBClassifier(
    n_estimators=400,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1,
    eval_metric="auc",
    feature_names=feature_names,
)

model.fit(x_train, y_train, eval_set=[(X_tuning, y_tuning)], verbose=False)

y_score_tuning = model.predict_proba(X_tuning)[:, 1]
y_score_held = model.predict_proba(X_held)[:, 1]

y_pred_tuning = (y_score_tuning >= 0.5).astype(int)
y_pred_held = (y_score_held >= 0.5).astype(int)

print("Tuning Accuracy:", accuracy_score(y_tuning, y_pred_tuning))
print("Held-out Accuracy:", accuracy_score(y_held, y_pred_held))

auroc_tuning = roc_auc_score(y_tuning, y_score_tuning)
auroc_held = roc_auc_score(y_held, y_score_held)

print("Tuning AUROC:", auroc_tuning)
print("Held-out AUROC:", auroc_held)
```

```python
import polars as pl

# One row per patient
patients = (
    df_input
    .group_by("Patient_ID", "Outcome")
    .first()
    .select("Patient_ID", "Outcome")
)

# Smallest class size
min_size = (
    patients
    .group_by("Outcome")
    .len()
    .select(pl.col("len").min())
    .item()
)

# Balanced patient subset
balanced_patients = (
    patients
    .group_by("Outcome")
    .map_groups(
        lambda g: g.sample(
            n=min_size,
            shuffle=True,
            seed=42
        )
    )
)

# Filter original dataframe
df_input = df_input.join(
    balanced_patients.select("Patient_ID"),
    on="Patient_ID",
    how="inner"
)
```

```python
import polars as pl
import numpy as np
from datetime import datetime, timedelta

start_min = datetime(2020, 1, 1)
start_max = datetime(2023, 1, 1)

def random_date():
    delta = start_max - start_min
    return start_min + timedelta(days=np.random.randint(delta.days))

base_dates = (
    df_input.select("Patient_ID")
      .unique()
      .with_columns(
          pl.col("Patient_ID").map_elements(lambda _: random_date()).alias("base_ts")
      )
)

df_input = (
    df_input.join(base_dates, on="Patient_ID")
      .with_columns(
          (
              pl.col("base_ts")
              + pl.duration(days=pl.col("Relative_Time"))
          ).alias("Timestamp")
      )
      .drop("base_ts")
)

df_input = df_input.rename({"lengthOfStay": "Length_of_Stay"})
```

```python
import polars as pl
from tqdm import tqdm
from rdflib import Graph
import os

from NEUROVASC2.sphn_mapping import (
    SEMANTIC_MAP,
    SemanticEvent,
    make_id,
    parse_code,
    render_event,
)

ROOT = "NEUROVASC2/MEDS_cohort"

ROWS_PER_FILE = 100_000

df = (
    pl.scan_parquet(f"{ROOT}/data/**/0.parquet")
    .select(["subject_id", "code", "time", "numeric_value"])
    .collect(engine="streaming")
)

total_rows = len(df)

file_index = 0
rows_in_file = 0

NUM_PATIENTS = len(df.group_by(pl.col("subject_id")).len())

OUTPUT_FOLDER = f"exports/neurovasc_v2/graph/sphn_pc_{NUM_PATIENTS}"

shutil.rmtree(OUTPUT_FOLDER)

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

f = open(f"{OUTPUT_FOLDER}/part_{file_index:03d}.nt", "w")

for row in tqdm(df.iter_rows(), total=total_rows, desc="Generating RDF"):
    if rows_in_file >= ROWS_PER_FILE:
        f.close()

        file_index += 1
        rows_in_file = 0

        f = open(f"{OUTPUT_FOLDER}/part_{file_index:03d}.nt", "w")

    variable, value = parse_code(row[1])

    if variable not in SEMANTIC_MAP:
        continue

    meta = SEMANTIC_MAP[variable]

    time = row[2]

    event_id = make_id(
        row[0],
        variable,
        time,
        value,
        row[3],
    )

    event = SemanticEvent(
        patient_id=row[0],
        variable=variable,
        numeric_value=row[3],
        time=time,
        template_name=meta,
        code=row[1],
        label=variable,
        event_id=event_id,
    )

    ev = render_event(event)
    # ev = Graph().parse(data=ev, format="nt").serialize(format="nt")

    f.write(ev)

    rows_in_file += 1

f.close()
```
